```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
import torch
```

### data loading

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.shape
```

### train test split

```python
# X = Features (independent variables), y = Target (what you want to predict)
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0],
    test_size=0.2,       # 20% allocated to the test set
    random_state=42,     # Ensures the split is reproducible
    stratify=df.iloc[:,0]         # Keeps the balance of target classes consistent
)
```

### scaling


```python
scaler = StandardScaler()

# 3. Fit on the training data ONLY and transform it
X_train = scaler.fit_transform(X_train)

# 4. Transform the test data using the training parameters (Do NOT refit!)
X_test = scaler.transform(X_test)
```

### Label Encoding

```python
y_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
print(y_train)
```

### Numpy arrays to PyTorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
y_train_tensor.shape
```

### Define the neural network

```python
class MySimpleNN():

  def __init__(self, X):

    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_calculation(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss



```

### Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

### Training Pipeline

```python
model = MySimpleNN(X_train)
```

```python
model.weights
```

```python
for epoch in range(epochs):
  #forward pass
  y_pred = model.forward(X_train_tensor)
  # calculate loss
  loss = model.loss_calculation(y_pred, y_train_tensor)

  # backpropogation
  loss.backward()

  # parameter update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')



```

```python
model.bias
```

### Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.85).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
